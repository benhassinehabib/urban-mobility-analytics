```python
# Create a copy for machine learning preparation
dc_ml = dc_clean.copy()
seoul_ml = seoul_clean.copy()

# Remove unnecessary columns from Washington DC
dc_ml = dc_ml.drop(columns=['instant', 'casual', 'registered'])

print("Washington DC remaining columns:")
print(dc_ml.columns.tolist())
```

```python
# Convert date columns to datetime format
dc_ml['date'] = pd.to_datetime(dc_ml['date'])
seoul_ml['date'] = pd.to_datetime(seoul_ml['date'])

# Extract useful time features
dc_ml['day_of_week'] = dc_ml['date'].dt.weekday
seoul_ml['day_of_week'] = seoul_ml['date'].dt.weekday

dc_ml['is_weekend'] = (dc_ml['day_of_week'] >= 5).astype(int)
seoul_ml['is_weekend'] = (seoul_ml['day_of_week'] >= 5).astype(int)

print("Washington DC - New features:")
print(dc_ml[['date', 'day_of_week', 'is_weekend']].head())

print("\nSeoul - New features:")
print(seoul_ml[['date', 'day_of_week', 'is_weekend']].head())
```

```python
# Extract month and year from date

dc_ml['month_from_date'] = dc_ml['date'].dt.month
dc_ml['year_from_date'] = dc_ml['date'].dt.year

seoul_ml['month_from_date'] = seoul_ml['date'].dt.month
seoul_ml['year_from_date'] = seoul_ml['date'].dt.year

# Display the new features
print("Washington DC:")
print(dc_ml[['date', 'month_from_date', 'year_from_date']].head())

print("\nSeoul:")
print(seoul_ml[['date', 'month_from_date', 'year_from_date']].head())
```

```python
# Create season feature from month

def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

dc_ml["season_from_date"] = dc_ml["month_from_date"].apply(get_season)
seoul_ml["season_from_date"] = seoul_ml["month_from_date"].apply(get_season)

# Display the new season feature
print("Washington DC:")
print(dc_ml[["date", "month_from_date", "season_from_date"]].head())

print("\nSeoul:")
print(seoul_ml[["date", "month_from_date", "season_from_date"]].head())
```

```python
# Encode categorical features using One-Hot Encoding

dc_ml = pd.get_dummies(
    dc_ml,
    columns=["season_from_date"],
    drop_first=True,
    dtype=int
)

seoul_ml = pd.get_dummies(
    seoul_ml,
    columns=["holiday", "functioning_day", "season_from_date"],
    drop_first=True,
    dtype=int
)

print("Washington DC columns:")
print(dc_ml.columns.tolist())

print("\nSeoul columns:")
print(seoul_ml.columns.tolist())
```

```python
# Check dataset sizes and target columns

print("Washington DC shape:", dc_ml.shape)
print("Seoul shape:", seoul_ml.shape)

print("\nWashington DC target:")
print(dc_ml["bike_rentals"].describe())

print("\nSeoul target:")
print(seoul_ml["bike_rentals"].describe())
```

```python
# Sort both datasets chronologically

dc_ml = dc_ml.sort_values(["date", "hour"]).reset_index(drop=True)
seoul_ml = seoul_ml.sort_values(["date", "hour"]).reset_index(drop=True)

# Separate predictors (X) and target (y)

X_dc = dc_ml.drop(columns=["bike_rentals", "date"])
y_dc = dc_ml["bike_rentals"]

X_seoul = seoul_ml.drop(columns=["bike_rentals", "date"])
y_seoul = seoul_ml["bike_rentals"]

# Check the results

print("Washington DC:")
print("X shape:", X_dc.shape)
print("y shape:", y_dc.shape)

print("\nSeoul:")
print("X shape:", X_seoul.shape)
print("y shape:", y_seoul.shape)
```

```python
# Split the data chronologically into training and testing sets

# Washington DC - 80% training, 20% testing
dc_split = int(len(X_dc) * 0.8)

X_dc_train = X_dc.iloc[:dc_split]
X_dc_test = X_dc.iloc[dc_split:]

y_dc_train = y_dc.iloc[:dc_split]
y_dc_test = y_dc.iloc[dc_split:]


# Seoul - 80% training, 20% testing
seoul_split = int(len(X_seoul) * 0.8)

X_seoul_train = X_seoul.iloc[:seoul_split]
X_seoul_test = X_seoul.iloc[seoul_split:]

y_seoul_train = y_seoul.iloc[:seoul_split]
y_seoul_test = y_seoul.iloc[seoul_split:]


# Check the sizes
print("Washington DC:")
print("Training:", X_dc_train.shape, y_dc_train.shape)
print("Testing:", X_dc_test.shape, y_dc_test.shape)

print("\nSeoul:")
print("Training:", X_seoul_train.shape, y_seoul_train.shape)
print("Testing:", X_seoul_test.shape, y_seoul_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Create the Linear Regression models
dc_model = LinearRegression()
seoul_model = LinearRegression()

# Train the models
dc_model.fit(X_dc_train, y_dc_train)
seoul_model.fit(X_seoul_train, y_seoul_train)

print("Models trained successfully!")
```

```python
# Make predictions on the test data

dc_predictions = dc_model.predict(X_dc_test)
seoul_predictions = seoul_model.predict(X_seoul_test)

print("Predictions generated successfully!")
print("Washington DC predictions:", dc_predictions[:10])
print("Seoul predictions:", seoul_predictions[:10])
```

```python
# Evaluate Linear Regression models

# Washington DC
dc_mae = mean_absolute_error(y_dc_test, dc_predictions)
dc_rmse = np.sqrt(mean_squared_error(y_dc_test, dc_predictions))
dc_r2 = r2_score(y_dc_test, dc_predictions)

# Seoul
seoul_mae = mean_absolute_error(y_seoul_test, seoul_predictions)
seoul_rmse = np.sqrt(mean_squared_error(y_seoul_test, seoul_predictions))
seoul_r2 = r2_score(y_seoul_test, seoul_predictions)

# Display results
print("WASHINGTON DC - Linear Regression")
print(f"MAE:  {dc_mae:.2f}")
print(f"RMSE: {dc_rmse:.2f}")
print(f"R²:   {dc_r2:.4f}")

print("\nSEOUL - Linear Regression")
print(f"MAE:  {seoul_mae:.2f}")
print(f"RMSE: {seoul_rmse:.2f}")
print(f"R²:   {seoul_r2:.4f}")
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Washington DC
axes[0].scatter(y_dc_test, dc_predictions, alpha=0.4)
axes[0].set_title("Actual vs Predicted - Washington DC")
axes[0].set_xlabel("Actual Bike Rentals")
axes[0].set_ylabel("Predicted Bike Rentals")

# Perfect prediction reference line
axes[0].plot(
    [y_dc_test.min(), y_dc_test.max()],
    [y_dc_test.min(), y_dc_test.max()],
    "r--"
)

# Seoul
axes[1].scatter(y_seoul_test, seoul_predictions, alpha=0.4)
axes[1].set_title("Actual vs Predicted - Seoul")
axes[1].set_xlabel("Actual Bike Rentals")
axes[1].set_ylabel("Predicted Bike Rentals")

axes[1].plot(
    [y_seoul_test.min(), y_seoul_test.max()],
    [y_seoul_test.min(), y_seoul_test.max()],
    "r--"
)

plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

# Create Random Forest models
dc_rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

seoul_rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Train the models
dc_rf_model.fit(X_dc_train, y_dc_train)
seoul_rf_model.fit(X_seoul_train, y_seoul_train)

print("Random Forest models trained successfully!")
```

```python
# Generate predictions using Random Forest models

dc_rf_predictions = dc_rf_model.predict(X_dc_test)
seoul_rf_predictions = seoul_rf_model.predict(X_seoul_test)

print("Random Forest predictions generated successfully!")
print("Washington DC:", dc_rf_predictions[:10])
print("Seoul:", seoul_rf_predictions[:10])
```

```python
# Evaluate Random Forest models

# Washington DC
dc_rf_mae = mean_absolute_error(y_dc_test, dc_rf_predictions)
dc_rf_rmse = np.sqrt(mean_squared_error(y_dc_test, dc_rf_predictions))
dc_rf_r2 = r2_score(y_dc_test, dc_rf_predictions)

# Seoul
seoul_rf_mae = mean_absolute_error(y_seoul_test, seoul_rf_predictions)
seoul_rf_rmse = np.sqrt(mean_squared_error(y_seoul_test, seoul_rf_predictions))
seoul_rf_r2 = r2_score(y_seoul_test, seoul_rf_predictions)

print("WASHINGTON DC - Random Forest")
print(f"MAE: {dc_rf_mae:.2f}")
print(f"RMSE: {dc_rf_rmse:.2f}")
print(f"R²: {dc_rf_r2:.4f}")

print("\nSEOUL - Random Forest")
print(f"MAE: {seoul_rf_mae:.2f}")
print(f"RMSE: {seoul_rf_rmse:.2f}")
print(f"R²: {seoul_rf_r2:.4f}")
```
